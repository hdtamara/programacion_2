## 10. Filtrado de datos — Boolean Indexing

Podemos filtrar elementos de un array aplicando una **condición booleana**. NumPy evalúa la condición para cada elemento y devuelve solo los que son `True`:

```python
arr[arr > 50]                     # elementos mayores a 50
arr[(arr > 30) & (arr < 60)]      # AND  →  &
arr[(arr < 20) | (arr > 80)]      # OR   →  |
```

> La condición crea un array de `True`/`False` (llamado **máscara booleana**) del mismo tamaño que el original. Luego se usa como índice.

> ⚠️ Con NumPy se usa `&` y `|` en lugar de `and` y `or`.

### `np.vectorize()` — función elemento a elemento

Cuando la lógica tiene muchos `if/elif`, anidar `np.where` se vuelve difícil de leer. `np.vectorize()` convierte cualquier función Python normal en una que opera elemento a elemento sobre un array.

```python
def clasificar(x):
    if x >= 90:  return 'Excelente'
    elif x >= 75: return 'Bueno'
    else:          return 'Regular'

clasificar_v = np.vectorize(clasificar)
clasificar_v(calif_np)   # aplica clasificar() a cada elemento
```

> ⚠️ `np.vectorize` es más **legible** que `np.where` anidado, pero no más rápida — internamente sigue iterando. Para rendimiento máximo usa `np.where`.

**¿Cuándo usar cada una?**

| Herramienta | Cuándo usarla |
|---|---|
| Boolean indexing `arr[cond]` | Filtrar — devuelve solo los elementos que cumplen |
| `np.where(cond, a, b)` | Asignar 2 valores según condición |
| `np.where` anidado | 3-4 categorías con lógica simple y lineal |
| `np.vectorize(func)` | Lógica compleja con múltiples `if/elif` |

## 12. El concepto de Axis (ejes)

El parámetro `axis` controla **en qué dirección** se aplica una función de agregación:

```
          columnas →
         [América, Europa, Asia, Resto]
filas ↓  [ 5.2,    3.8,   2.1,  0.9  ]   ← juego 0
         [ 3.5,    4.2,   6.8,  1.2  ]   ← juego 1
         ...                              ...
```

| `axis` | Dirección | Resultado |
|---|---|---|
| `axis=0` | Colapsa filas ↓ | Un valor **por columna** (por región) |
| `axis=1` | Colapsa columnas → | Un valor **por fila** (por juego) |

```python
ventas_np.sum(axis=0)   # total vendido en cada región (suma todas las filas)
ventas_np.sum(axis=1)   # total vendido por cada juego (suma todas las columnas)
ventas_np.mean(axis=1)  # promedio de ventas por regiones para cada juego
```

## 13. Reshape, Transpose y Flatten

Estas operaciones cambian la **forma** de un array sin modificar sus datos:

| Método | Qué hace |
|---|---|
| `.reshape(filas, cols)` | Cambia la forma (el número total de elementos debe ser igual) |
| `.T` | Transpone: intercambia filas y columnas |
| `.flatten()` | Convierte cualquier array a 1D (una sola fila) |

```python
arr = np.arange(12)           # [0, 1, 2, ..., 11]  →  shape (12,)
arr.reshape(3, 4)             # 3 filas, 4 columnas  →  shape (3, 4)

ventas_np.flatten()           # todos los 40 valores en un vector
```

## 14. Broadcasting

**Broadcasting** es la capacidad de NumPy para operar entre arrays de formas diferentes, extendiendo automáticamente el array más pequeño.

```python
# ventas_np tiene shape (10, 4)
# ventas_totales tiene shape (10,)

# Para dividir cada fila por su total necesitamos convertir ventas_totales
# a shape (10, 1) → así NumPy "extiende" la división columna por columna
porcentajes = ventas_np / ventas_totales[:, np.newaxis] * 100
#                                          ↑
#                         convierte (10,) en (10, 1)
```

## 15. Ordenamiento — `np.sort()` y `np.argsort()`

| Función | Qué devuelve |
|---|---|
| `np.sort(arr)` | Array con los **valores** ordenados (de menor a mayor) |
| `np.argsort(arr)` | Array con los **índices** que ordenarían el array |

`argsort()` es especialmente útil porque permite ordenar **otros arrays en paralelo**:

```python
orden = np.argsort(calif_np)           # índices de menor a mayor calificación
orden_desc = np.argsort(calif_np)[::-1]  # índices de mayor a menor

nombres_np[orden_desc]    # nombres ordenados por calificación
calif_np[orden_desc]      # calificaciones en ese mismo orden
```
